```python
import shap
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(input_df)
print(shap_values)
```

```python
# === FULL SHAP PIPELINE (CORRECTED) ===

import numpy as np

# STEP 1: Prepare input (VERY IMPORTANT)
input_sample = input_df.iloc[0:1]
input_sample = input_sample[X_train.columns].astype(float)

# STEP 2: Get prediction index SAFELY
probs = model.predict_proba(input_sample)[0]
y_pred = probs.argmax()   # ensures valid index

# STEP 3: Compute SHAP values
shap_values = explainer.shap_values(
    input_sample,
    #check_additivity=False
)

# STEP 4: Handle both SHAP formats safely
if isinstance(shap_values, list):
    # old format
    shap_values_for_pred = shap_values[y_pred][0]
else:
    # new format (your case: 3D array)
    shap_values_for_pred = shap_values[0, :, y_pred]

# STEP 5: Map with feature names
feature_names = X_train.columns
results = list(zip(feature_names, shap_values_for_pred))

# STEP 6: Remove very small values (noise)
results = [r for r in results if abs(r[1]) > 0.01]

# STEP 7: Sort by importance
results_sorted = sorted(results, key=lambda x: abs(x[1]), reverse=True)
top_features = results_sorted[:10]

# STEP 8: Separate supporting vs opposing
supporting = [f for f, v in top_features if v > 0][:5]
opposing = [f for f, v in top_features if v < 0][:3]

# STEP 9: Print final output
print("\n=== PREDICTION EXPLANATION ===\n")


print(f"Predicted class index: {le.inverse_transform([y_pred])[0]}")

print("\nTop contributing symptoms:")
for f in supporting:
    print(f"✔ {f}")

print("\nOpposing symptoms:")
for f in opposing:
    print(f"✖ {f}")

print("\nDetailed contributions:")
for feature, value in top_features:
    direction = "supports" if value > 0 else "against"
    print(f"{feature}: {value:.4f} ({direction})")
```
