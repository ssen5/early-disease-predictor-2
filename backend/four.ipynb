```python
import joblib
import pandas as pd

model = joblib.load("edpmodel.pkl")
le = joblib.load("edple.pkl")
```

```python
from base_dict_2 import base_dict
from meaning import getMeaning

s1= ["burning_urination", "frequent_urination" , "abdominal_pain"]
s2= ["excessive_thirst","frequent_urination", "slow_healing_wounds", "blurred vision"]
s3= ["high_fever" , "chills" , "night_sweats" , "body_ache"]
s4= ["high_fever" ,"neck_stiffness" , "confusion" , "headache"]
s5= ["pale_stool","dizziness"]
s6= ['vomiting', 'diarrhea', 'constipation']
s7= ["runny_nose","shortness_of_breath","dry_cough"]
 
symps = s5

inp_base = base_dict()

for symp in symps:
    if symp in inp_base:
        inp_base[symp] = 1

input_df = pd.DataFrame([inp_base])


y_pred = model.predict(input_df)

decoded_label = le.inverse_transform([y_pred[0]])

print("Predicted Disease:", decoded_label[0])

getMeaning(decoded_label[0])
```
